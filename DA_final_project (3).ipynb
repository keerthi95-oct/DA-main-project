**Stage 2 – Data Cleaning and Pre-processing**

**Text cleaning**

```python
category_cols = ["Region", "Product", "StoreLocation", "CustomerType", "Salesperson", "PaymentMethod", "RegionManager"]
for col in category_cols:
    df[col] = df[col].str.strip().str.title()
    print(col, df[col].unique())
```

**Handling missing value**

```python
df["Promotion"] = df["Promotion"].fillna("No Promotion")
```

```python
df["Promotion"].isnull().sum()
df["Promotion"].value_counts()
```

**Removed duplicates**

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

**Date Conversion**

```python
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
df["OrderDate"] = pd.to_datetime(df["OrderDate"], errors="coerce")
```

**Creating Date Features**

```python
df["Year"] = df["Date"].dt.year
df["Month"] = df["Date"].dt.month
df["Month_Name"] = df["Date"].dt.month_name()
df["Day_Name"] = df["Date"].dt.day_name()
```

```python
df[["Date", "Year", "Month", "Month_Name", "Day_Name"]].head()
```

**Removed extra spaces**

```python
text_columns = df.select_dtypes(include="object").columns

for col in text_columns:
    df[col] = df[col].str.strip()
```

**Calculating Delivery Days**

```python
df['DeliveryDays'] = (df['DeliveryDate'] - df['OrderDate']).dt.days
```

```python
df[['OrderDate', 'DeliveryDate', 'DeliveryDays']].head()
```

**Creating Discount Level**

```python
df["Discount_Level"] = (df["Discount"] * 100).astype(int).astype(str) + "%"
```

```python
df[["Discount_Level"]].head()
```

**Creating Order Size**

```python
df["Order_Size"] = pd.cut(df["TotalPrice"], [0, 1000, 3000, 6000, np.inf],
                          labels=["Small", "Medium", "Large", "Bulk"])
```

```python
df[["Order_Size"]].head()
```

**Creating Net Revenue**

```python
df["NetRevenue"] = df["TotalPrice"] * (1 - df["Returned"])
```

```python
df[["NetRevenue"]].head()
```

**Region Manager check**

```python
df.groupby("Region")["RegionManager"].nunique().max()
```

**Outlier**

```python
num_cols = ["Quantity", "UnitPrice", "Discount", "TotalPrice", "ShippingCost", "DeliveryDays"]

for col in num_cols:
    Q1, Q3 = df[col].quantile([0.25, 0.75])
    IQR = Q3 - Q1
    n = ((df[col] < Q1 - 1.5 * IQR) | (df[col] > Q3 + 1.5 * IQR)).sum()
    print(f"{col}: outliers = {n}, skewness = {df[col].skew():.2f}")
```

Only TotalPrice had outliers (17), with a moderate right skew of 0.97.

```python
plt.figure(figsize=(8, 5))
sns.histplot(df["TotalPrice"], kde=True, color="steelblue")
plt.axvline(df["TotalPrice"].mean(), color="red", linestyle="--", label="Mean")
plt.axvline(df["TotalPrice"].median(), color="green", linestyle="--", label="Median")
plt.title("Distribution of TotalPrice")
plt.legend()
plt.show()
```

A long tail to the right means positive skew.
The red mean line sits to the right of the green median line, because large orders pull the average up.
In a symmetrical chart, the two lines would be almost on top of each other.

# **Stage 3 – EDA and Visualizations**

**Chart 1 – Sales by Region**





```python
region_sales = df.groupby("Region")["TotalPrice"].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))
region_sales.plot(kind="bar")

plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

**Chart Type:** Bar Chart

**Analysis Type:** Descriptive Analysis


**Interpretation:**




The bar chart compares total sales across different regions.

It shows which region generated the highest and lowest sales.

The Region column is used for categories, while TotalPrice is used to calculate total sales.

This helps identify regional sales performance.

**Insights:**

 North has the highest sales and South the lowest. The other three regions are close together.

**Key Insights**

Returns are a major revenue concern – Around 24.8% of orders were returned, resulting in approximately 1.1M in lost revenue. Cash payments and Chairs show particularly high return rates.

Higher discounts do not increase order value – Average order value decreases from approximately 3,067 at 0% discount to 2,686 at 15% discount, indicating that deeper discounts may not be effective.

Free shipping performs strongly – FREESHIP generated the highest net revenue (935K) compared with percentage-based promotions such as SAVE10.

Large orders contribute most of the revenue – Although Small orders occur frequently, Large and Bulk orders contribute significantly more revenue.

Sales have no consistent monthly trend – Sales fluctuate over time without a steady pattern. Revenue increased only slightly from 1.26M in 2023 to 1.35M in 2024.

Regional and product performance varies – North performs strongly, while South needs improvement. Tablet, Laptop and Printer are among the strongest products, while Phone has weaker sales.

**Future Enhancements**

Optimize promotions by increasing the use of free-shipping offers and reviewing the effectiveness of SAVE10 and WINTER15.

Reduce product returns by investigating the reasons behind returns, especially for cash payments and high-return products such as Chairs.

Improve weaker regions and products through targeted marketing campaigns, especially for South, West and Phone.

Increase larger orders by introducing product bundles, bulk-order incentives, or minimum-order offers.

Improve sales performance across stores and salespeople by identifying successful practices from high-performing stores and salespeople and applying them to weaker areas.

Improve delivery performance by reducing the current average delivery time of approximately 6.04 days.
